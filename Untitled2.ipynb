```python
!pip install yfinance pandas matplotlib
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Apple hissesinin verisini çekiyoruz
data = yf.download("AAPL", start="2025-08-19", end="2026-08-19")

# Çektiğimiz verinin ilk 5 satırını ekrana yazdıralım
print(data.head())
```

```python
# 1. Hareketli Ortalama ve Günlük Getiri Hesaplama
data['MA_20'] = data['Close'].rolling(window=20).mean()
data['Daily_Return'] = data['Close'].pct_change()

# 2. Fiyat ve Trend Grafiği Çizimi
plt.figure(figsize=(12, 6))
plt.plot(data['Close'], label='Kapanış Fiyatı', alpha=0.5)
plt.plot(data['MA_20'], label='20 Günlük Hareketli Ortalama', color='red')
plt.title("AAPL Hisse Fiyatı ve Trend Analizi")
plt.xlabel("Tarih")
plt.ylabel("Fiyat ($)")
plt.legend()
plt.grid(True)
plt.show()

# 3. İstatistiksel Özet Tablosu
print("\n--- İstatistiksel Özet ---")
print(data['Close'].describe())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Verimiz zaten yukarıda 'data' değişkeninde duruyor.
# Tarihleri sayısal bir indekse çevirelim ki model anlayabilsin.
df = data.reset_index()
df['Day_Index'] = np.arange(len(df))

# 2. X (Girdi: Günler) ve y (Çıktı: Kapanış Fiyatları) olarak ayıralım
X = df[['Day_Index']]
y = df['Close']

# 3. Modelimizi kuralım ve eğitelim (Fit edelim)
model = LinearRegression()
model.fit(X, y)

# 4. Modelin tahmin ettiği çizgiyi çizdirelim
df['Prediction'] = model.predict(X)

# 5. Görselleştirme
plt.figure(figsize=(12, 6))
plt.plot(df['Close'], label='Gerçek Fiyat', color='blue', alpha=0.5)
plt.plot(df['Prediction'], label='İstatistiksel Trend Çizgisi', color='red', linewidth=2)
plt.title('AAPL Hisse Fiyatı ve Basit Trend Analizi')
plt.xlabel('Zaman (Günler)')
plt.ylabel('Fiyat ($)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error

# Modelin tahmin ettiği fiyatlar ile gerçek fiyatlar arasındaki ortalama farkı bulalım
hata = mean_absolute_error(df['Close'], df['Prediction'])

print(f"Modelimizin ortalama hata payı (MAE): ${hata:.2f}")
```

```python
# Mevcut gün sayısının üzerine 5 gün daha ekleyelim
gelecek_gunler = np.arange(len(df), len(df) + 5).reshape(-1, 1)

# Gelecek günler için fiyat tahmini yapalım
gelecek_fiyatlar = model.predict(gelecek_gunler)

print("Önümüzdeki 5 gün için tahmin edilen fiyatlar:")
for i, fiyat in enumerate(gelecek_fiyatlar, 1):
    # Fiyatı tek bir sayıya (float) çevirerek yazdırıyoruz
    print(f"{i}. Gün Sonraki Tahmini Fiyat: ${float(fiyat):.2f}")
```
